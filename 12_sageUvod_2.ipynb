**Pazite!**


```sage
floor(1/(2.1-2))
```

Ampak
$$ \left\lfloor \frac{1}{2.1 - 2} \right\rfloor = \left\lfloor \frac{1}{.1} \right\rfloor = \left\lfloor 10 \right\rfloor = 10$$

Kaj se je zgodilo?


Računalniki shranjujejo realna števila v binarni obliki, mi pa smo navajeni uporabljati desetiško predstavitev. Število $2.1$ v desetiškem zapisu je precej preprosto in kratko, vendar je po pretvorbi v binarni zapis $10.0\overline{0011} = 10.0001100110011 \dots$


Ker računalniki ne morejo shraniti neskončnega števila številk, se to število nekje zaokroži, kar povzroči manjšo napako, ki smo jo videli.
V SageMathu pa so racionalna števila (ulomki) natančna, zato te napake pri zaokroževanju nikoli ne bomo videli.


```sage
floor(1/ ( (21/10) - 2) )
```

# Pripisovanje, enakost, enačbe in neenačbe.


SageMath uporablja `=` za pripisovanje. Za primerjavo uporablja `==`, `<=`, `>=`, `<` in `>`:


```sage
a=5
a
```

```sage
2 == 2
```

```sage
2<=3
```

```sage
2<2
```

```sage
a < 3
```

```sage
a == 5
```

Če želimo shraniti rezultat izračuna, ga dodelimo spremenljivki:


```sage
y=1+2
```

**Pazite**: Pogosta napaka je uporaba izraza `a=b` za primerjavo `a` in `b`, vendar na koncu vrednost `a` zamenjate z vrednostjo `b`.


```sage
y*(y+2)
```

Rezultat izračuna se ne natisne samodejno, ko je pripisan spremenljivki.
Zato bomo naredili naslednje, da ga tudi natisnemo,


```sage
b=3*4 ; b
```

Znak `;`, ki ločuje več navodil v isti vrstici.
Če ni določeno drugače, se v beležnici Jupyter natisne samo izpis, ki ga ustvari zadnja vrstica. Zato je uporaba znaka `;` enakovredna uporabi nove vrstice v isti celici.


```sage
y = 3 + 2; y
y+2
```

Če želimo natisniti vsak korak, moramo uporabiti ukaz `print()`


```sage
y = 1 + 2; print(y)
y = 3 * y + 1; print(y)
y = 3 * y + 1; print(y)
y = 3 * y + 1; print(y)
```

- Če znate programirati v Pythonu, boste morda ugotovili, da je to standardno obnašanje spremenljivke.


- Ni priporočljivo na novo definirati vnaprej določenih konstant in funkcij iz programa Sage.
- Čeprav to ne vpliva na notranje obnašanje programa Sage, lahko privede do presenetljivih rezultatov:


```sage
pi = -I/2
```

```sage
exp(2*I*pi)
```

Ukaz `restore()` povrne privzeto vrednost vsem vnaprej določenim spremenljivkam in funkcijam.


```sage
restore()
```

```sage
y+pi
```

Ukaz `reset()` opravi še popolnejšo ponastavitev, zlasti izbriše vse uporabniško definirane spremenljivke.


```sage
reset()
```

```sage
y
```

Opazite tudi, da SageMath včasih razlikuje med simbolnimi izračuni in števili.


```sage
e^(2*pi*I)
```

```sage
e^(2*pi*I) == 1
```

SageMath zgornjega ne razume kot primerjavo, temveč kot enačbo.


```sage
3*x - 10 == 5
```

Enačbo ali neenakost rešimo z ukazom `solve()`.


```sage
solve(3*x - 2 == 5,x)
```

Privzeto SageMath interpretira `x` kot simbolno spremenljivko. Za zdaj bomo reševali samo `x`, pozneje pa bomo videli, kako določiti lastne (simbolne) spremenljivke


```sage
solve( 2*x - 5 >= 17,x)
```

```sage
solve( x^2 + x == 6, x)
```

```sage
solve(2*x^2 - x + 1 == 0, x)
```

Množica rešitev nekaterih neenačb je sestavljena iz unije in preseka odprtih intervalov.


```sage
solve( x^2 - 6 >= 3, x )
```

```sage
solve( x^2 - 6 <= 3, x )
```

- Ukaz `solve()` poskuša izraziti rešitev enačbe brez uporabe števil s plavajočo vejico.
- Če tega ni mogoče storiti, bo vrnil rešitev v simbolični obliki.


```sage
solve( sin(x) == x, x)
```

- Za iskanje numeričnega približka rešitve lahko uporabimo ukaz `find_root()`.
- Ta ukaz zahteva izraz in zaprt interval, na katerem se išče rešitev


```sage
find_root(sin(x) == x, -pi/2 , pi/2)
```

- Ob začetku seje SageMath ustvari eno simbolno spremenljivko, `x`, ki se lahko uporablja za reševanje enačb.
- Če želite uporabiti dodatno simbolno spremenljivko, jo morate deklarirati z ukazom `var()`.


```sage
y,z,t = var("y z t")
```

```sage
phi, theta, rho = var("phi theta rho")
```

```sage
x1, x2 = var("x1 x2")
```

- Imena spremenljivk ne smejo vsebovati presledkov
- Na primer `square root` ni veljavno ime spremenljivke, `square_root` pa je.


Poskus uporabe simbolne spremenljivke, preden je bila deklarirana, bo povzročil napako `NameError`.


```sage
u
```

```sage
solve (u^2-1,u)
```

Simbolno spremenljivko, kot je zgoraj opredeljena spremenljivka `phi`, lahko razglašamo z ukazom `restore()`.


```sage
restore('phi')
```

```sage
phi
```

Majhne sisteme linearnih enačb lahko rešite tudi s ukazom `solve()`, če imajo vse simbolne spremenljivke so bile deklarirane.


```sage
x,y,z = var('x','y','z')
```

```sage
solve( [3*x - y == 2,-2*x -y == 1 ], x,y)
```

```sage
solve( [ 2*x - y == -1 , 2*x - y == 2],x,y)
```

```sage
solve( [ 2*x + y == -1 ,-4*x - 2*y == 2],x,y)
```

- `r1` označuje, da obstaja prosta spremenljivka, ki določa množico rešitev.


- Če je več prostih spremenljivk, jih SageMath našteje `r1,r2,..., rk`.


```sage
solve([ 2*x + 3*y + 5*z == 1, 4*x + 6*y + 10*z == 2, 6*x + 9*y + 15*z == 3], x,y,z)
```

- Ukaz `solve()` je lahko pri velikih sistemih enačb zelo počasen.
- Za te sisteme je najbolje uporabiti funkcije linearne algebre, saj so precej učinkovite. Kmalu se bomo naučili, kako.


Reševanje neenačb v več spremenljivkah lahko privede do zapletenih izrazov, saj so območja, ki jih določajo, zapletena.


```sage
solve([ x-y >=2, x+y <=3], x,y)
```
